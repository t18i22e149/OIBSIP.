```python
# DATA CLEANING USING PYTHON (PANDAS)
# Netflix Dataset
# ============================================

import pandas as pd
import numpy as np

# Load Dataset
import pandas as pd

df = pd.read_csv(r"C:\Users\Admin\Downloads\OBSIP\Triveni_Task3\Dataset\netflix1.csv")

# Display First 5 Rows
print("Original Dataset:")
print(df.head())
```

```python
# 1. DATA INTEGRITY
# Ensuring accuracy, consistency, and reliability
# =========================================================

print("\nDataset Shape:", df.shape)

# Check Data Types
print("\nData Types:")
print(df.dtypes)

# Check Null Values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# 2. MISSING DATA HANDLING
# Filling missing values
# =========================================================

# Fill missing text columns with 'Unknown'
text_columns = ['director', 'country', 'rating']

for col in text_columns:
    df[col] = df[col].fillna("Unknown")

# Convert date column to datetime
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# Fill missing dates with forward fill
df['date_added'] = df['date_added'].fillna(method='ffill')

print("\nMissing Values After Handling:")
print(df.isnull().sum())
```

```python
# =========================================================
# 3. DUPLICATE REMOVAL
# Removing duplicate records
# =========================================================

# Check duplicates
print("\nDuplicate Rows:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

print("Shape After Removing Duplicates:", df.shape)
```

```python
# 4. STANDARDIZATION
# Making data format consistent
# =========================================================

# Convert titles to Title Case
df['title'] = df['title'].str.title()

# Remove extra spaces
df['director'] = df['director'].str.strip()
df['country'] = df['country'].str.strip()

# Standardize column names
df.columns = df.columns.str.lower()

# Convert duration to numeric only
df['duration_number'] = df['duration'].str.extract('(\d+)')
df['duration_number'] = pd.to_numeric(df['duration_number'])

print("\nStandardized Data:")
print(df.head())
```

```python
# 5. OUTLIER DETECTION
# Detecting unusual values using IQR method
# =========================================================

Q1 = df['release_year'].quantile(0.25)
Q3 = df['release_year'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Detect outliers
outliers = df[
    (df['release_year'] < lower_bound) |
    (df['release_year'] > upper_bound)
]

print("\nOutliers in Release Year:")
print(outliers[['title', 'release_year']])

# Remove outliers
df = df[
    (df['release_year'] >= lower_bound) &
    (df['release_year'] <= upper_bound)
]

print("\nShape After Removing Outliers:", df.shape)
```

```python
# FINAL CLEANED DATASET
# =========================================================

print("\nFinal Cleaned Dataset:")
print(df.head())

# Save Cleaned Dataset
df.to_csv("cleaned_netflix_dataset.csv", index=False)

print("\nCleaned dataset saved successfully!")
```
