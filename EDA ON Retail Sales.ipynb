```python
#Step 1 — Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Step 2 – Load the Dataset

df = pd.read_csv(r"C:\Users\Admin\Downloads\OBSIP\Triveni_Task 1\Dataset\retail_sales_dataset.csv")

print(df.head())
```

```python
# Step 3 — Understand the Dataset

print(df.info())
```

```python
print(df.isnull().sum())
```

```python
#Remove Duplicates
df.drop_duplicates(inplace=True)
```

```python
#Convert Date Column
df['Date'] = pd.to_datetime(df['Date'])
```

```python
#Basic stastices
print(df.describe())
```

```python
# Sales analysis 
print(df['Total Amount'].sum())
```

```python
#Time Series anlaysis 
monthly_sales = df.groupby(df['Date'].dt.month)['Total Amount'].sum()

monthly_sales.plot(kind='line', marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
#Product Analysis 
product_sales = df.groupby('Product Category')['Total Amount'].sum()

print(product_sales)
```

```python
#Bar chart
product_sales.plot(kind='bar')
plt.title("Sales by Product Category")
plt.ylabel("Sales")
plt.show()
```

```python
#Customer analysis 
gender_sales = df.groupby('Gender')['Total Amount'].sum()

gender_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title("Gender-wise Sales")
plt.show()
```

```python
#Corelation heat map
numeric_df = df.select_dtypes(include=np.number)

sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
#Advanced Visualizations Histogram
df['Total Amount'].hist(bins=20)
plt.title("Sales Distribution")
plt.show()
#Boxplot
sns.boxplot(x=df['Total Amount'])
plt.title("Outlier Detection")
plt.show()
```

```python
#Pairplot
sns.pairplot(df)
plt.show()
#Countplot
# Countplot

sns.countplot(x='Product Category', data=df)

plt.xticks(rotation=45)

plt.show()
```
