```python
# Fraud Detection using Machine Learning
# Dataset: creditcard.csv

# STEP 1 - Import Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Libraries Imported Successfully")


# STEP 2 - Load Dataset
df = pd.read_csv("C:/Users/Admin/Downloads/OBSIP/Triveni_Task7/Dataset/creditcard.csv")

print("\nDataset Loaded Successfully")
print(df.head())


# STEP 3 - Dataset Information
print("\nDataset Information")
print(df.info())


# STEP 4 - Check Missing Values
print("\nMissing Values")
print(df.isnull().sum())


# STEP 5 - Check Fraud and Normal Transactions
print("\nTransaction Counts")
print(df['Class'].value_counts())

# 0 = Normal Transaction
# 1 = Fraud Transaction


# STEP 6 - Separate Features and Target
X = df.drop('Class', axis=1)
y = df['Class']

print("\nFeatures Shape :", X.shape)
print("Target Shape :", y.shape)


# STEP 7 - Feature Scaling
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("\nFeature Scaling Completed")
print(X_scaled[:5])


# STEP 8 - Split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)

print("\nData Split Completed")
print("Training Data :", X_train.shape)
print("Testing Data :", X_test.shape)


# STEP 9 - Train Machine Learning Model
model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)

print("\nModel Training Completed")


# STEP 10 - Prediction
y_pred = model.predict(X_test)

print("\nPrediction Completed")
print(y_pred[:20])


# STEP 11 - Accuracy Score
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy Score :", accuracy)


# STEP 12 - Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix")
print(cm)


# STEP 13 - Classification Report
report = classification_report(y_test, y_pred)

print("\nClassification Report")
print(report)


# STEP 14 - Test Custom Transaction
sample = X_test[0].reshape(1, -1)

prediction = model.predict(sample)

print("\nCustom Transaction Prediction")

if prediction[0] == 0:
    print("Normal Transaction")
else:
    print("Fraud Transaction")
```

```python
# STEP 15 - Import Visualization Libraries

import matplotlib.pyplot as plt


# BAR CHART - Fraud vs Normal Transactions

transaction_counts = df['Class'].value_counts()

plt.figure(figsize=(6,5))

plt.bar(['Normal', 'Fraud'], transaction_counts)

plt.title("Fraud vs Normal Transactions")
plt.xlabel("Transaction Type")
plt.ylabel("Count")

plt.show()



# PIE CHART - Fraud Distribution

plt.figure(figsize=(6,6))

plt.pie(
    transaction_counts,
    labels=['Normal', 'Fraud'],
    autopct='%1.2f%%'
)

plt.title("Transaction Distribution")

plt.show()



# HISTOGRAM - Transaction Amount Distribution

plt.figure(figsize=(8,5))

plt.hist(df['Amount'], bins=30)

plt.title("Transaction Amount Distribution")
plt.xlabel("Amount")
plt.ylabel("Frequency")

plt.show()



# LINE CHART - First 100 Transaction Amounts

plt.figure(figsize=(10,5))

plt.plot(df['Amount'][:100])

plt.title("First 100 Transaction Amounts")
plt.xlabel("Transaction Number")
plt.ylabel("Amount")

plt.show()
```
